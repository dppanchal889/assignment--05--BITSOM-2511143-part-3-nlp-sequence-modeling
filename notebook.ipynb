## Task 1: Dataset Understanding

```python
import pandas as pd
```

# Load dataset

```python
df = pd.read_csv("customer_support_text_classification.csv")
```

# 1. Number of records

```python
print("Number of records:", len(df))
```

# 2. Target labels/classes

```python
print("\nTarget labels:")
print(df["sentiment_label"].unique())
```

# 3. Sample text records

```python
print("\nSample customer messages:")
print(df["customer_message"].head())
```

# 4. Average text length

```python
text_lengths = df["customer_message"].apply(lambda x: len(x.split()))

average_length = text_lengths.mean()

print("\nAverage text length:", average_length)
```

# 5. Class distribution

```python
print("\nClass distribution:")
print(df["sentiment_label"].value_counts())
```
